```python
Category_sells_most = df.groupby('Category')['Sales'].sum()
Category_sells_most
```

```python
Most_profit_category = df.groupby('Category')['Profit'].sum().idxmax()
Most_profit_category
```

```python
region_high_profit = df.groupby('Region')['Profit'].sum().idxmax()
region_high_profit
```

```python
sales_change_overtime = df.groupby(df['Order Date'].dt.year)['Sales'].sum()
sales_change_overtime
```

```python
Higher_discount = df['Discount'].corr(df['Profit'])
Higher_discount
```

```python
category_sales = df.groupby('Category')['Sales'].sum()
category_sales
```

```python
category_sales.plot(kind='bar', figsize=(8,5))

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum()
category_profit
```

```python
category_profit.plot(kind='bar', figsize=(8,5))

plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.xticks(rotation=0)
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
region_sales
```

```python
monthly_sales = df.groupby('Year_Month')['Sales'].sum()
monthly_sales
```

```python
monthly_sales.plot(figsize=(12,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
top_products = (
    df.groupby('Product Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
discount_profit = df.groupby('Discount')['Profit'].mean()
discount_profit
```

```python
discount_profit.plot(kind='line', marker='o', figsize=(8,5))

plt.title('Average Profit by Discount')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.show()
```

```python
top_customers = (
    df.groupby('Customer Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```
